## Opdracht 2b: Opsplitsen taak 2: afleiding ontwerpeis

$$1/3=1/(1-\frac{\omega^2}{\omega_o^2})$$
$$\frac{\omega^2}{\omega_0^2} = 4$$
$$f_0 = 5/2 hz$$

```python
# Voeg hier je plot in.
def amplitude(f):
    1/(1-((f*2*np.pi)**2/(25*np.pi**2)))
    return 1/(1-((f*2*np.pi)**2/(25*np.pi**2)))

f_vloer = np.logspace(-1,1,1000)

A_out = np.abs(np.array(amplitude(f_vloer)))
plt.figure()
plt.plot(f_vloer, A_out)
plt.xlabel('Vloerfrequentie ($hz$)')
plt.ylabel('Amplitudeoverdracht')
plt.xscale('log')
plt.yscale('log')
plt.show()

```

```python
g=9.81
omega_0 = 5*np.pi

delta_x = g/omega_0**2
print(delta_x)
```

## Opdracht 3: Plot de grafiek.

```python
# Voeg hier je plot in. Lees goed de opdracht wat er in de grafiek moet staan!
# Het is ook handig om je grafiek op te slaan, doe dat door met plt.savefig().

def delta_x_param(m, k):
    c = (m*g)/(k*delta_x)
    return c

massa = np.linspace(0,5,100)
veerconstantes_1 = delta_x_param(massa, 1)
veerconstantes_1_5 = delta_x_param(massa, 1.5)
veerconstantes_2 = delta_x_param(massa, 2)
veerconstantes_5 = delta_x_param(massa, 5)

plt.figure()
plt.plot(massa,veerconstantes_1, label='k=1')
plt.plot(massa,veerconstantes_1_5, label='k=1.5')
plt.plot(massa,veerconstantes_2, label='k=2')
plt.plot(massa,veerconstantes_5, label='k=5')
plt.legend()
plt.grid()
plt.xlabel('Massa ($kg$)')
plt.ylabel('Veerconstante($N/m$)')
plt.show()
```
