Zie probleem 2.6 in het boek van Griffith.

1. Wat moet de hoogte $z$ zijn boven de schijf zodat het elektrische veld door de schijf 10% is in vergelijking met dat van een oneindig geladen plaat?

Voeg hieronder je analytische uitwerking toe.

##### Antwoord 2h

Het elektrische veld van een oneindig geladen plaat is:

$$
E_{\infty}=\frac{\sigma}{2\varepsilon_0}
$$

We zoeken de hoogte z waarbij het elektrisch veld door de schijf 10% is in vergelijking met dat van een oneindig geladen plaat:

$$
E_{\text{schijf}}(z)=0.10\,E_{\infty}
$$

Invullen geeft:

$$
\frac{\sigma}{2\varepsilon_0}\left(1-\frac{z}{\sqrt{R^2+z^2}}\right)
=
0.10\frac{\sigma}{2\varepsilon_0}
$$

Vereenvoudigen:

$$
\frac{z}{\sqrt{R^2+z^2}}=0.90
$$

z isoleren:

$$
z^2=\frac{0.81}{0.19}R^2=\frac{81}{19}R^2
$$

$$
z=\sqrt{\frac{81}{19}}\,R=\frac{9}{\sqrt{19}}\,R
$$

Numeriek:

$$
z \approx 2.06\,R
$$

Om ervoor te zorgen dat het elektrische veld van de schijf 10% is van dat van een oneindig geladen plaat, moet gelden: 

$$
z \approx 2.06\,R
$$

#### Vraag 3c: De oppervlaktelading van een geleidende schijf op constant potentiaal

De oppervlakte lading $\sigma(r)$ van een geleidende schijf met straal $R$ is niet uniform verdeeld over de schijf, maar wordt gegeven door (Problems 2.58 and 2.63 Griffiths 5th edition):
$$\sigma(r) = \frac{Q}{2 \pi R \sqrt{R^2 - r^2}}$$

1. Bereken de totale lading op de schijf met behulp van vraag 3a en 3b.
2. Vergelijk de oppervlakte lading in het centrum van de geleidende schijf met dat van een uniform geladen schijf met dezelfde totale lading.
3. Plot de oppervlakte lading van de geleidende schijf met de berekende lading $Q$, tussen $r = 0$ en $R$.
4. Plot in hetzelfde figuur de oppervlaktelading van een uniform geladen schijf met dezelfde lading $Q$.


````python
# 1
R = disk_radius
Ez0_plus = Ez[idx_above, 0]
sigma0 = epsilon0 * Ez0_plus

Q = 2*np.pi*R**2*sigma0

print("3c.1 Totale lading op de schijf:")
print(f"  Q  = {Q:.6e} C")

# 2
sigma_uniform = Q / (np.pi * R**2)

print("\n3c.2 Vergelijking met uniforme schijf:")
print(f"  sigma_uniform = {sigma_uniform:.6e} C/m^2")
print(f"  sigma_center (geleidend) = {sigma0:.6e} C/m^2")
print(f"  verhouding sigma_uniform / sigma_center = {sigma_uniform/sigma0:.3f}")

# 3 en 4
r_plot = np.linspace(0, R*(1 - 1e-6), 600)

sigma_conducting = Q / (2 * np.pi * R * np.sqrt(R**2 - r_plot**2))
sigma_uniform_curve = np.full_like(r_plot, sigma_uniform)

plt.figure()
plt.title("Oppervlakteladingsdichtheid σ(r) op een schijf")
plt.plot(r_plot, sigma_conducting, label=r"Geleidend: $\sigma(r)=\frac{Q}{2\pi R\sqrt{R^2-r^2}}$")
plt.plot(r_plot, sigma_uniform_curve, label=r"Uniform: $\sigma=\frac{Q}{\pi R^2}$")
plt.xlabel("r (m)")
plt.ylabel(r"$\sigma$ (C/m$^2$)")
plt.grid(True)
plt.legend()
plt.show()
````
